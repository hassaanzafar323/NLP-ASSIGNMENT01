```python
import pandas as pd
import os

# Data load function for both train and val
def load_data():
    train_path = '../data/train.csv'
    val_path = '../data/val.csv'
    
    if os.path.exists(train_path) and os.path.exists(val_path):
        train_df = pd.read_csv(train_path)
        val_df = pd.read_csv(val_path)
        print(f"Loaded Train Dataset: {len(train_df)} rows")
        print(f"Loaded Validation Dataset: {len(val_df)} rows")
        return train_df, val_df
    else:
        print("Dataset files not found in '../data/' folder!")
        return None, None

train_df, val_df = load_data()
```

```python
if train_df is not None:
    print(" TRAIN DATASET BASIC STATISTICS ")
    print(f"Total Documents (Rows): {len(train_df)}")
    
    # Average document length in words (assuming column name is 'text')
    if 'text' in train_df.columns:
        train_df['word_count'] = train_df['text'].astype(str).apply(lambda x: len(x.split()))
        print(f"Average Document Length: {train_df['word_count'].mean():.2f} words")
    
    
    target_col = None
    for col in train_df.columns:
        if col.lower() in ['target', 'label', 'sentiment', 'category']:
            target_col = col
            break
            
    if target_col:
        print(f"\nTarget Distribution ({target_col}):")
        print(train_df[target_col].value_counts())
    else:
        print("\nNote: No standard target/label column detected automatically.")
```

```python
if train_df is not None and 'text' in train_df.columns:
    
    all_tokens = []
    for text in train_df['text'].dropna().astype(str):
        all_tokens.extend(text.split())
        
    total_tokens = len(all_tokens)
    vocabulary_size = len(set(all_tokens))
    
    print("VOCABULARY & STATS ")
    print(f"Total Tokens (Word Count): {total_tokens}")
    print(f"Vocabulary Size (Unique Types): {vocabulary_size}")
    print(f"Type-Token Ratio (TTR): {vocabulary_size / total_tokens:.4f}")
    
    print("\n Zipf's Law Observation")
    print("Observation: A small number of words (like stop words, common punctuation) appear very frequently,")
    print("while a massive portion of the vocabulary consists of rare words (long tail distribution),")
    print("which is a classic characteristic aligned with Zipf's Law in NLP corpora.")
```

```python
if train_df is not None and 'text' in train_df.columns:
    print("=== REAL MESSY TEXT EXAMPLES FROM DATASET ===")
    messy_samples = train_df['text'].dropna().head(5).tolist()
    
    for idx, sample in enumerate(messy_samples, 1):
        print(f"Example {idx}:")
        print(f"{sample}\n" + "-"*40)
```
