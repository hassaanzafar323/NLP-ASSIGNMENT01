```python
import pandas as pd
import os

train_path = '../data/train.csv'
val_path = '../data/val.csv'

if os.path.exists(train_path) and os.path.exists(val_path):
    train_df = pd.read_csv(train_path, header=None, names=['id', 'topic', 'sentiment', 'text']).dropna(subset=['text'])
    val_df = pd.read_csv(val_path, header=None, names=['id', 'topic', 'sentiment', 'text']).dropna(subset=['text'])
    
    print(f"Train Shape: {train_df.shape} | Val Shape: {val_df.shape}")
    print(f"Total Documents: {len(train_df)}")
    
    train_df['word_count'] = train_df['text'].astype(str).apply(lambda x: len(x.split()))
    print(f"Avg Doc Length: {train_df['word_count'].mean():.2f} words")
    
    print("\nSentiment Distribution:")
    print(train_df['sentiment'].value_counts())
    
    all_tokens = [word for text in train_df['text'].astype(str) for word in text.split()]
    print(f"\nTotal Tokens: {len(all_tokens)}")
    print(f"Vocabulary Size: {len(set(all_tokens))}")
    print(f"TTR: {len(set(all_tokens)) / len(all_tokens):.4f}")
    
    print("\n=== ZIPF'S LAW OBSERVATION ===")
    print("Observation: A small number of words (like stop words, punctuation, and common terms)")
    print("appear extremely frequently in the corpus, while a massive portion of the vocabulary")
    print("consists of rare words that appear only a few times (heavy long-tail distribution).")
    print("This inverse relationship between frequency and rank is a classic characteristic of Zipf's Law in NLP.")
    
    print("\n=== MESSY TEXT EXAMPLES ===")
    for idx, sample in enumerate(train_df['text'].head(5), 1):
        print(f"Example {idx}: {sample}")
```
